```python
import pandas as pd
import numpy as np
import random

# 1. Creación de datos "sucios" para el ejercicio
data = {
    'ID_Estudiante': [101, 102, 103, 104, 105, 106, 102, 108, 109, 110], # 102 está duplicado
    'Nombre': ['Ana', 'Luis', 'Carlos', 'MARIA', 'jorge', np.nan, 'Luis', 'Sofia', 'Pedro', 'Lucia'], # Formatos inconsistentes y nulos
    'Edad': [20, 22, 19, 21, 500, 23, 22, 20, 24, 21], # 500 es un valor atípico (outlier)
    'Promedio': [4.5, 3.8, np.nan, 4.0, 3.5, 4.2, 3.8, 4.9, 2.5, 3.0], # Valor faltante
    'Semestre': ['Sem-1', 'Sem-2', 'Sem-1', 'Sem-2', 'Sem-1', 'Sem-3', 'Sem-2', 'Sem-1', 'Sem-1', 'Sem-1']
}

df_sucio = pd.DataFrame(data)
# Guardamos esto como si fuera el archivo que "descargaste"
df_sucio.to_csv('datos_estudiantes_sucios.csv', index=False)

print("Archivo CSV sucio creado exitosamente.")
```

```python

# Cargar librerías necesarias
import pandas as pd
import numpy as np
from sklearn.preprocessing import MinMaxScaler

# 1. Cargar los datos
df = pd.read_csv('datos_estudiantes_sucios.csv')

print("--- Primeras filas del Dataset Original ---")
print(df.head())
print("\n--- Información General ---")
print(df.info())
```

```python
# 2. LIMPIEZA

# A. Manejo de duplicados
duplicados = df.duplicated().sum()
print(f"\nDuplicados encontrados: {duplicados}")
df = df.drop_duplicates() # Eliminar duplicados

# B. Manejo de valores faltantes (Imputación)
# Rellenamos el 'Promedio' faltante con la media de la columna
promedio_media = df['Promedio'].mean()
df['Promedio'] = df['Promedio'].fillna(promedio_media)

# Eliminamos filas donde el 'Nombre' sea nulo (ya que es difícil imputar un nombre)
df = df.dropna(subset=['Nombre'])

# C. Manejo de valores atípicos (Outliers)
# Detectamos edades imposibles (ej. mayores a 100 años)
filtro_outliers = df['Edad'] <= 100
df = df[filtro_outliers] # Nos quedamos solo con los datos válidos

print("\n--- Dataset después de Limpieza ---")
print(df)
```

```python
# 3. TRANSFORMACIÓN

# A. Formatear datos categóricos
# Poner todos los nombres en formato Título (primera mayúscula)
df['Nombre'] = df['Nombre'].str.title().str.strip()

# B. Crear nuevas columnas derivadas
# Categorizar: Si promedio >= 4.0 es 'Alto', sino 'Básico'
df['Categoria_Notas'] = np.where(df['Promedio'] >= 4.0, 'Alto', 'Básico')

# C. Normalización de columnas numéricas
# Vamos a escalar el 'Promedio' para que esté entre 0 y 1
scaler = MinMaxScaler()
df['Promedio_Normalizado'] = scaler.fit_transform(df[['Promedio']])

print("\n--- Dataset Transformado ---")
print(df.head())
```

```python
# 4. EXPORTACIÓN
df.to_csv('datos_estudiantes_limpios.csv', index=False)
print("\n¡Archivo final exportado con éxito: 'datos_estudiantes_limpios.csv'!")
```
