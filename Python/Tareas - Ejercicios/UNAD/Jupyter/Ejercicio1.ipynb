```python
import numpy as np

# 1. Crear dos matrices 3x3 con valores aleatorios.
# np.random.rand(3, 3) genera valores entre 0 y 1.
matriz_A = np.random.rand(3, 3)
matriz_B = np.random.rand(3, 3)

print("--- Matrices Originales ---")
print("Matriz A:\n", matriz_A)
print("\nMatriz B:\n", matriz_B)

# 2. Realizar las operaciones solicitadas.
# Suma de matrices. [cite: 18]
suma_matrices = matriz_A + matriz_B

# Resta de matrices. [cite: 22]
resta_matrices = matriz_A - matriz_B

# Producto matricial (producto punto). [cite: 23]
producto_matricial = np.dot(matriz_A, matriz_B)
# Alternativa: producto_matricial = matriz_A @ matriz_B

# Transposición de la Matriz A. [cite: 24]
transpuesta_A = matriz_A.T

# 3. Mostrar los resultados de cada operación. 
print("\n\n--- Resultados de las Operaciones ---")
print("Suma (A + B):\n", suma_matrices)
print("\nResta (A - B):\n", resta_matrices)
print("\nProducto Matricial (A * B):\n", producto_matricial)
print("\nTranspuesta de A:\n", transpuesta_A)
```
