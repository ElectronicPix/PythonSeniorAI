```python
import numpy as np

# 1. Generar una matriz 4x4 con valores aleatorios entre 1 y 100. [cite: 28]
matriz_original = np.random.randint(1, 101, size=(4, 4))

def transformar_matriz(matriz):
    """
    Identifica los elementos > 50 y los convierte en 1, el resto en 0. [cite: 29]
    Utilizamos np.where para una implementación eficiente y vectorizada.
    Sintaxis: np.where(condición, valor_si_verdadero, valor_si_falso)
    """
    matriz_transformada = np.where(matriz > 50, 1, 0)
    return matriz_transformada

# 2. Aplicar la función a la matriz original.
matriz_final = transformar_matriz(matriz_original)

# 3. Mostrar la matriz original y la transformada. 
print("--- Matriz Original (valores de 1 a 100) ---")
print(matriz_original)

print("\n--- Matriz Transformada (valores > 50 son 1, el resto 0) ---")
print(matriz_final)
```
