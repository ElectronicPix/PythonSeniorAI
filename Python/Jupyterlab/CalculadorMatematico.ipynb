```python
import sympy
from sympy import symbols, Eq, solve, sin, cos, tan, exp
import matplotlib.pyplot as plt
import numpy as np

# --- 1. Definir símbolos ---
x, y, z = symbols('x y z')

# --- 2. Cálculo de Derivadas e Integrales  ---
funcion_f = x**3 + cos(x)

# Derivada
derivada_f = sympy.diff(funcion_f, x)
print(f"La función es: {funcion_f}")
print(f"La derivada es: {derivada_f}")

# Integral Indefinida
integral_f = sympy.integrate(funcion_f, x)
print(f"La integral es: {integral_f}")

# --- 3. Resolución de Ecuaciones Lineales (3 variables) [cite: 68] ---
# Ejemplo:
# ec1: x + y + z = 1
# ec2: x - y + z = 5
# ec3: 2x + y - z = 0

ec1 = Eq(x + y + z, 1)
ec2 = Eq(x - y + z, 5)
ec3 = Eq(2*x + y - z, 0)

solucion = solve((ec1, ec2, ec3), (x, y, z))
print(f"La solución del sistema es: {solucion}")

# --- 4. Graficación de Funciones [cite: 69] ---
print("Graficando funciones...")

# Definimos el rango para el eje X
x_vals = np.linspace(-2 * np.pi, 2 * np.pi, 400) # Intervalo de -2pi a 2pi

# Convertimos funciones de SymPy a funciones que NumPy pueda usar
f_sin = sympy.lambdify(x, sin(x), 'numpy')
f_cos = sympy.lambdify(x, cos(x), 'numpy')
f_tan = sympy.lambdify(x, tan(x), 'numpy')
f_exp = sympy.lambdify(x, exp(x), 'numpy')

# Creamos las gráficas
plt.figure(figsize=(12, 8))

plt.subplot(2, 2, 1)
plt.plot(x_vals, f_sin(x_vals), label='sin(x)')
plt.title("Función Seno")
plt.grid(True)
plt.legend()

plt.subplot(2, 2, 2)
plt.plot(x_vals, f_cos(x_vals), label='cos(x)', color='r')
plt.title("Función Coseno")
plt.grid(True)
plt.legend()

plt.subplot(2, 2, 3)
# Ajustamos el rango Y para la tangente
y_tan = f_tan(x_vals)
y_tan[y_tan > 10] = np.nan # Evitar picos infinitos
y_tan[y_tan < -10] = np.nan
plt.plot(x_vals, y_tan, label='tan(x)', color='g')
plt.title("Función Tangente")
plt.ylim(-10, 10)
plt.grid(True)
plt.legend()

plt.subplot(2, 2, 4)
plt.plot(x_vals, f_exp(x_vals), label='exp(x)', color='purple')
plt.title("Función Exponencial")
plt.grid(True)
plt.legend()

plt.tight_layout() # Ajusta el layout
plt.show()
```
